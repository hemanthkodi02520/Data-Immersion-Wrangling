```python
# Load dataset
import pandas as pd
df = pd.read_csv(r"C:\Users\heman\OneDrive\Tài liệu\customer_data.csv")
print("Dataset Shape:", df.shape)
```

```python
# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
# Remove duplicates
df.drop_duplicates(inplace=True)

# Fill missing Age with median
df["Age"].fillna(df["Age"].median(), inplace=True)

# Fill missing PurchaseAmount with mean
df["PurchaseAmount"].fillna(df["PurchaseAmount"].mean(), inplace=True)
```

```python
# Standardize Gender values
df["Gender"] = df["Gender"].str.strip().str.title()

# Convert JoinDate to datetime
df["JoinDate"] = pd.to_datetime(df["JoinDate"])
```

```python
# Save cleaned dataset
df.to_csv("customer_data_cleaned.csv", index=False)

print("\nCleaning Completed Successfully!")
print("Cleaned Dataset Saved as customer_data_cleaned.csv")
```

```python
#Complete Program
import pandas as pd
df = pd.read_csv(r"C:\Users\heman\OneDrive\Tài liệu\customer_data.csv")
print("Dataset Shape:", df.shape)
print("\nMissing Values:")
print(df.isnull().sum())
df.drop_duplicates(inplace=True)
df["Age"].fillna(df["Age"].median(), inplace=True)
df["PurchaseAmount"].fillna(df["PurchaseAmount"].mean(), inplace=True)
df["Gender"] = df["Gender"].str.strip().str.title()
df["JoinDate"] = pd.to_datetime(df["JoinDate"])
df["AgeGroup"] = pd.cut(
    df["Age"],
    bins=[0, 20, 30, 40, 60],
    labels=["Teen", "Young Adult", "Adult", "Senior"]
)
df.to_csv("customer_data_cleaned.csv", index=False)

print("\nCleaning Completed Successfully!")
print("Cleaned Dataset Saved as customer_data_cleaned.csv")
```
